### 8.12 Outlier Treatment

```python
# Check outliers using IQR

numeric_columns = df_clean.select_dtypes(include='number').columns

for column in numeric_columns:

    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print(column, ":", len(outliers), "outliers")
```

```python
# Example: inspect SALES outliers

Q1 = df_clean['SALES'].quantile(0.25)
Q3 = df_clean['SALES'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

sales_outliers = df_clean[
    (df_clean['SALES'] < lower_limit) |
    (df_clean['SALES'] > upper_limit)
]

print(sales_outliers[['ORDERNUMBER', 'PRODUCTLINE', 'QUANTITYORDERED', 'SALES']])
```

### Outlier Treatment

Outliers were identified using the Interquartile Range (IQR) method. The detected extreme values were inspected before deciding whether to remove them.

Sales and quantity outliers may represent legitimate large customer orders rather than data-entry errors. Therefore, outliers were not removed automatically. Only values confirmed to be erroneous would be treated or removed.

### 9.1 Create a working dataset

```python
# Create a working copy for data processing

df_processed = df_clean.copy()

print("Dataset shape:")
print(df_processed.shape)
```

```python
# Check ORDERDATE

print(df_processed['ORDERDATE'].dtype)
print(df_processed['ORDERDATE'].head())
```

```python
# Extract year from ORDERDATE Annual sales

df_processed['YEAR'] = df_processed['ORDERDATE'].dt.year

print(df_processed[['ORDERDATE', 'YEAR']].head())
```

```python
# Extract month number

df_processed['MONTH'] = df_processed['ORDERDATE'].dt.month

print(df_processed[['ORDERDATE', 'MONTH']].head())
```

```python
# Extract month name

df_processed['MONTH_NAME'] = df_processed['ORDERDATE'].dt.month_name()

print(df_processed[['ORDERDATE', 'MONTH_NAME']].head())
```

```python
# Create quarter from order date

df_processed['QUARTER'] = df_processed['ORDERDATE'].dt.quarter

print(df_processed[['ORDERDATE', 'QUARTER']].head())
```

### 9.4 Feature Engineering

```python
# Calculate revenue per unit

df_processed['REVENUE_PER_UNIT'] = (
    df_processed['SALES'] / df_processed['QUANTITYORDERED']
)

print(
    df_processed[
        ['SALES', 'QUANTITYORDERED', 'REVENUE_PER_UNIT']
    ].head()
)
```

### 9.5 Feature Selection

```python
# Select important columns for sales analysis

sales_data = df_processed[
    [
        'ORDERNUMBER',
        'ORDERDATE',
        'YEAR',
        'MONTH',
        'MONTH_NAME',
        'QUARTER',
        'PRODUCTLINE',
        'QUANTITYORDERED',
        'PRICEEACH',
        'SALES',
        'COUNTRY',
        'DEALSIZE',
        'STATUS'
    ]
]

print(sales_data.head())
print("Shape:", sales_data.shape)
```

### 9.6 Sorting

```python
# Sort records by order date

df_processed = df_processed.sort_values(
    by='ORDERDATE'
)

print(
    df_processed[
        ['ORDERDATE', 'ORDERNUMBER', 'SALES']
    ].head(10)
)
```

### 9.7 Grouping and Aggregation — Sales by Product Line

```python
# Calculate total sales by product line

productline_sales = (
    df_processed
    .groupby('PRODUCTLINE')['SALES']
    .sum()
    .sort_values(ascending=False)
)

print(productline_sales)
```

### 9.8 Sales by Country

```python
# Calculate total sales by country

country_sales = (
    df_processed
    .groupby('COUNTRY')['SALES']
    .sum()
    .sort_values(ascending=False)
)

print(country_sales)
```

### 9.9 Sales by Deal Size

```python
# Calculate sales by deal size

deal_size_sales = (
    df_processed
    .groupby('DEALSIZE')['SALES']
    .sum()
    .sort_values(ascending=False)
)

print(deal_size_sales)
```

### 9.10 Sales by Year

```python
# Calculate yearly sales

yearly_sales = (
    df_processed
    .groupby('YEAR')['SALES']
    .sum()
    .sort_index()
)

print(yearly_sales)
```

### 9.11 Sales by Month

```python
# Calculate monthly sales

monthly_sales = (
    df_processed
    .groupby('MONTH')['SALES']
    .sum()
    .sort_index()
)

print(monthly_sales)
```

### 9.12 Sales by Year and Month

```python
# Calculate sales by year and month

monthly_sales_trend = (
    df_processed
    .groupby(['YEAR', 'MONTH'])['SALES']
    .sum()
    .reset_index()
)

print(monthly_sales_trend.head(15))
```

### 9.13 Quantity Sold by Product Line

```python
# Calculate total quantity ordered by product line

product_quantity = (
    df_processed
    .groupby('PRODUCTLINE')['QUANTITYORDERED']
    .sum()
    .sort_values(ascending=False)
)

print(product_quantity)
```

### 9.14 Average Sales by Product Line

```python
# Calculate average sales per transaction by product line

average_product_sales = (
    df_processed
    .groupby('PRODUCTLINE')['SALES']
    .mean()
    .sort_values(ascending=False)
)

print(average_product_sales)
```

### 9.15 Filter Completed/Successful Orders

```python
# Filter shipped orders

shipped_orders = df_processed[
    df_processed['STATUS'] == 'Shipped'
]

print(shipped_orders.head())
print("Number of shipped records:", len(shipped_orders))
```

### 9.16 Filter Large Deals

```python
# Filter large deals

large_deals = df_processed[
    df_processed['DEALSIZE'] == 'Large'
]

print(large_deals.head())
print("Number of large deals:", len(large_deals))
```

### 9.17 Create a Product Performance Table

```python
# Create product performance summary

product_performance = (
    df_processed
    .groupby('PRODUCTLINE')
    .agg(
        Total_Sales=('SALES', 'sum'),
        Average_Sales=('SALES', 'mean'),
        Total_Quantity=('QUANTITYORDERED', 'sum'),
        Average_Price=('PRICEEACH', 'mean')
    )
    .sort_values(
        by='Total_Sales',
        ascending=False
    )
)

print(product_performance)
```

### 9.18 Create Country Performance Table

```python
# Create country performance summary

country_performance = (
    df_processed
    .groupby('COUNTRY')
    .agg(
        Total_Sales=('SALES', 'sum'),
        Average_Sales=('SALES', 'mean'),
        Total_Quantity=('QUANTITYORDERED', 'sum'),
        Number_of_Orders=('ORDERNUMBER', 'nunique')
    )
    .sort_values(
        by='Total_Sales',
        ascending=False
    )
)

print(country_performance)
```

```python
# Identify approximate peak location using histogram bins

for column in ['SALES', 'QUANTITYORDERED', 'PRICEEACH']:
    
    data = df_processed[column].dropna()
    
    counts, bins = np.histogram(data, bins=30)
    
    peak_bin = counts.argmax()
    
    print(
        column,
        "peak range:",
        round(bins[peak_bin], 2),
        "to",
        round(bins[peak_bin + 1], 2)
    )
```

## Peak Analysis

### Objective

To identify the ranges where numerical observations are most concentrated.

### Observation

The analysis identifies the histogram range containing the largest number of observations for selected numerical variables.

### Interpretation

A peak indicates a range where observations occur more frequently than in surrounding ranges.

### Finding

The peak ranges provide an indication of the most common values or value ranges within the selected numerical variables.

```python
# Identify potential extreme values using the IQR method

outlier_summary = {}

for column in numeric_columns:
    
    data = df_processed[column].dropna()
    
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    extreme_values = data[
        (data < lower_limit) |
        (data > upper_limit)
    ]
    
    outlier_summary[column] = len(extreme_values)

print("Potential Extreme Values:")
print(
    pd.Series(outlier_summary)
    .sort_values(ascending=False)
)
```

### Objective

The Interquartile Range (IQR) method was used. Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were considered potential extreme values.

### Interpretation

Potential extreme values do not necessarily represent errors. They may represent legitimate transactions, such as unusually large orders or high-value sales.

### Finding

The number of potential extreme values was calculated for each numerical variable. These observations should be investigated before deciding whether any treatment is necessary.
